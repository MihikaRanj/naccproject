# NACC Horizon & Etiology Counts (Exploration Notebook)

This notebook is **only for counting** incident dementia cases by follow-up horizon (3y, 4y, 5y, …) and breaking them down by **etiology subtype** (AD / PD_LBD / Vascular / Other).

It **does not** write `cohort.csv`.


## 0) Setup

Update the file paths below to point to your NACC visit-level dataset (the one that has repeated visits per `NACCID`).


```python
import pandas as pd
import numpy as np
import os

pd.set_option('display.max_columns', 200)
pd.set_option('display.width', 200)

DATA_DIR = os.path.join("..", "data")

DATA_PATH = os.path.join(DATA_DIR, "investigator_nacc72.csv")

# If your file has different date column names, update here:
NACCID_COL   = "NACCID"
VISIT_DATE_COL = "visit_date"   # or e.g., "VISDATE" depending on your extract
```

## 1) Load data

This assumes your dataset is **long format** (multiple rows per person over time).


```python
df = pd.read_csv(DATA_PATH, low_memory=False)
print(df.shape)
df.head()
```

## 2) Parse dates and define baseline per participant

Baseline is defined as each participant’s **first observed visit** (minimum visit date).
If you already have a baseline table, you can swap this section out.


```python
# Ensure date type
df[VISIT_DATE_COL] = pd.to_datetime(df[VISIT_DATE_COL], errors='coerce')

# Drop rows with missing keys/dates
df = df.dropna(subset=[NACCID_COL, VISIT_DATE_COL]).copy()

# Baseline = first visit per person
baseline = (df.sort_values([NACCID_COL, VISIT_DATE_COL])
              .groupby(NACCID_COL, as_index=False)
              .first()
              .rename(columns={VISIT_DATE_COL: 'baseline_date'}))

print("n participants:", baseline[NACCID_COL].nunique())
baseline[[NACCID_COL, 'baseline_date']].head()
```

## 3) Utility: time difference in years


```python
def years_between(a: pd.Series, b: pd.Series) -> pd.Series:
    return (b - a).dt.days / 365.25
```

## 4) Join baseline dates to all visits and compute follow-up time


```python
df2 = df.merge(baseline[[NACCID_COL, 'baseline_date']], on=NACCID_COL, how='left').copy()
df2['t_years'] = years_between(df2['baseline_date'], df2[VISIT_DATE_COL])

# post-baseline visits only
post = df2[df2['t_years'] > 0].copy()

# max observed follow-up per participant
max_follow = post.groupby(NACCID_COL)['t_years'].max().rename('max_follow_years')
baseline = baseline.merge(max_follow, on=NACCID_COL, how='left')
baseline['max_follow_years'] = baseline['max_follow_years'].fillna(0)

baseline[['max_follow_years']].describe()
```

## 5) Identify first dementia event per participant

We treat dementia as:
- If `NACCUDSD` exists: dementia event if `NACCUDSD == 4`
- Else fallback: if `DEMENTED == 1`

If your extract uses a different code/column, update the logic here.


```python
if 'NACCUDSD' in post.columns:
    dem_events = post[post['NACCUDSD'] == 4].copy()
    dementia_flag_used = 'NACCUDSD==4'
elif 'DEMENTED' in post.columns:
    dem_events = post[post['DEMENTED'] == 1].copy()
    dementia_flag_used = 'DEMENTED==1'
else:
    raise ValueError("No dementia indicator found. Expected NACCUDSD or DEMENTED.")

print("Dementia event definition:", dementia_flag_used)
print("dementia-event rows:", dem_events.shape)

# first dementia event per participant
first_dem = (dem_events.sort_values([NACCID_COL, VISIT_DATE_COL])
                     .groupby(NACCID_COL, as_index=False)
                     .first()
                     .rename(columns={VISIT_DATE_COL: 'dementia_date'}))

# attach baseline_date for time-to-event
first_dem = first_dem.merge(baseline[[NACCID_COL, 'baseline_date']], on=NACCID_COL, how='left')
first_dem['time_to_dem_years'] = years_between(first_dem['baseline_date'], first_dem['dementia_date'])

print("Participants with any post-baseline dementia event:", first_dem[NACCID_COL].nunique())
first_dem[[NACCID_COL, 'baseline_date', 'dementia_date', 'time_to_dem_years']].head()
```

## 6) Etiology subtype mapping (customizable)

Below is a **simple, explicit rule-set** based on common NACC variables.
Adjust the rules to match your project’s definition.

Currently:
- **PD_LBD** if `NACCLBDE == 1`
- **Vascular** if `CVD == 1` (or if `CVDIF` indicates VBI etiology)
- **AD** if `NACCALZD == 1` (if present)
- else **Other/Unk**

If you want to use your hierarchical labels (Stage2 target), replace `assign_subtype()` accordingly.


```python
def assign_subtype(row) -> str:
    # --- PD/LBD ---
    if 'NACCLBDE' in row and pd.notna(row['NACCLBDE']) and int(row['NACCLBDE']) == 1:
        return 'PD_LBD'

    # --- Vascular brain injury / vascular dementia proxy ---
    if 'CVD' in row and pd.notna(row['CVD']) and int(row['CVD']) == 1:
        return 'Vascular'
    if 'CVDIF' in row and pd.notna(row['CVDIF']) and int(row['CVDIF']) in {1, 2, 3}:
        return 'Vascular'

    # --- AD ---
    if 'NACCALZD' in row and pd.notna(row['NACCALZD']) and int(row['NACCALZD']) == 1:
        return 'AD'

    return 'Other/Unk'

first_dem['subtype'] = first_dem.apply(assign_subtype, axis=1)
first_dem['subtype'].value_counts(dropna=False)
```

## 7) Horizon sweep: cumulative dementia counts by year

This prints cumulative incident dementia cases by horizon (1y, 2y, …). 
It also reports how many participants even have **at least** that much follow-up.


```python
max_year = int(np.ceil(first_dem['time_to_dem_years'].max())) if len(first_dem) else 0
max_year = max(max_year, int(np.ceil(baseline['max_follow_years'].max())))
max_year = min(max_year, 20)  # safety cap; adjust if you want

rows = []
for y in range(1, max_year + 1):
    # people with at least y years of observed follow-up
    eligible = (baseline['max_follow_years'] >= y).sum()
    # people with dementia by y years (regardless of having >=y follow-up)
    dem_by_y = (first_dem['time_to_dem_years'] <= y).sum()
    # people with dementia by y years AND had >=y follow-up (stricter)
    dem_by_y_strict = ((first_dem['time_to_dem_years'] <= y) &
                      (first_dem[NACCID_COL].isin(baseline.loc[baseline['max_follow_years'] >= y, NACCID_COL]))).sum()
    rows.append((y, eligible, dem_by_y, dem_by_y_strict))

horizon_table = pd.DataFrame(rows, columns=['horizon_years', 'n_with_>=horizon_followup', 'n_dementia_by_horizon', 'n_dementia_by_horizon_and_>=followup'])
horizon_table
```

## 8) Horizon sweep: dementia counts by subtype

For each horizon, this shows how many **first dementia events** fall within that time window, by subtype.


```python
subtype_horizons = []
for y in range(1, max_year + 1):
    temp = first_dem[first_dem['time_to_dem_years'] <= y]
    vc = temp['subtype'].value_counts()
    subtype_horizons.append({
        'horizon_years': y,
        'AD': int(vc.get('AD', 0)),
        'PD_LBD': int(vc.get('PD_LBD', 0)),
        'Vascular': int(vc.get('Vascular', 0)),
        'Other/Unk': int(vc.get('Other/Unk', 0)),
        'Total': int(len(temp))
    })

subtype_table = pd.DataFrame(subtype_horizons)
subtype_table
```

## 9) Quick view: marginal gain per year

This helps you spot when increasing horizon stops adding many PD_LBD / Vascular cases.


```python
marginal = subtype_table.copy()
for col in ['AD','PD_LBD','Vascular','Other/Unk','Total']:
    marginal[col + '_delta'] = marginal[col].diff().fillna(marginal[col]).astype(int)

marginal[['horizon_years',
          'PD_LBD','PD_LBD_delta',
          'Vascular','Vascular_delta',
          'Total','Total_delta']]
```
